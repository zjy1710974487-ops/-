```python
# ============================================================
# Shared utilities for:
#   fig:refit_mismatch
#   fig:fisher_bias
#   fig:window_scan
#
# Save as, e.g., emri_transition_figures.py
# ============================================================

import os
from dataclasses import dataclass, replace

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid
from scipy.interpolate import interp1d
from numpy.linalg import inv, eig

# -------------------------------
# plotting style
# -------------------------------
plt.rcParams.update({
    "figure.dpi": 140,
    "savefig.dpi": 300,
    "font.size": 11,
    "axes.labelsize": 11,
    "axes.titlesize": 11,
    "axes.linewidth": 1.0,
    "lines.linewidth": 1.8,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 4,
    "ytick.major.size": 4,
    "xtick.minor.size": 2,
    "ytick.minor.size": 2,
    "legend.frameon": False,
})

# -------------------------------
# physical constants
# -------------------------------
G = 6.67430e-11
c = 299792458.0
MSUN = 1.988409870698051e30
PC = 3.085677581491367e16

MTSUN_SI = G * MSUN / c**3


@dataclass
class Params:
    # benchmark from the draft
    M: float = 2.0e5
    mu: float = 1.4
    a: float = 0.90

    # observational setup
    DL_Gpc: float = 1.0
    f_start: float = 0.0030
    f_cut: float = 0.0364

    # mixed-phase window
    f_h: float = 0.012
    f_q: float = 0.021

    # transition sector
    Lambda_H: float = 450.0
    Lambda_Q: float = 5.0

    # phenomenological kernel
    # these are tunable because your manuscript fixes the structure
    # but not a unique final coefficient set
    kappa0: float = 2.0e6
    h2: float = 0.0
    h3s: float = 0.0
    h4: float = 0.0

    # numerics
    nfreq: int = 4000
    match_tc_days: float = 0.05
    ntc: int = 1201

    # finite differences for Fisher
    eps_lnM: float = 2.0e-3
    eps_lnmu: float = 2.0e-3
    eps_a: float = 2.0e-4


# ============================================================
# unit helpers
# ============================================================
def M_sec(M_msun: float) -> float:
    return M_msun * MTSUN_SI


def D_sec(D_Gpc: float) -> float:
    D_m = D_Gpc * 1.0e9 * PC
    return D_m / c


# ============================================================
# Kerr circular equatorial dynamics
# ============================================================
def v_from_f22(f22, p: Params):
    return np.power(np.pi * M_sec(p.M) * f22, 1.0 / 3.0)


def f22_from_v(v, p: Params):
    return v**3 / (np.pi * M_sec(p.M))


def omega_from_v(v, p: Params):
    return v**3 / M_sec(p.M)


def u_of_v(v, a):
    return v**2 / np.power(1.0 - a * v**3, 2.0 / 3.0)


def E_tilde(v, a):
    u = u_of_v(v, a)
    N = 1.0 - 2.0 * u + a * u**1.5
    D = 1.0 - 3.0 * u + 2.0 * a * u**1.5
    return N / np.sqrt(D)


def dEtilde_dv(v, a):
    u = u_of_v(v, a)
    N = 1.0 - 2.0 * u + a * u**1.5
    D = 1.0 - 3.0 * u + 2.0 * a * u**1.5
    Np = -2.0 + 1.5 * a * np.sqrt(u)
    Dp = -3.0 + 3.0 * a * np.sqrt(u)
    dEdu = (2.0 * Np * D - N * Dp) / (2.0 * D**1.5)
    dudv = 2.0 * v / np.power(1.0 - a * v**3, 5.0 / 3.0)
    return dEdu * dudv


# ============================================================
# phenomenological transition sector
# ============================================================
def Fhat_PP(v, a):
    return (
        1.0
        - (1247.0 / 336.0) * v**2
        + (4.0 * np.pi - (73.0 / 12.0) * a) * v**3
        - (44711.0 / 9072.0) * v**4
    )


def smoothstep(x):
    y = np.zeros_like(x)
    mid = (x >= 0.0) & (x <= 1.0)
    y[x > 1.0] = 1.0
    y[mid] = 3.0 * x[mid]**2 - 2.0 * x[mid]**3
    return y


def Lambda_of_v(v, p: Params):
    vh = v_from_f22(p.f_h, p)
    vq = v_from_f22(p.f_q, p)
    x = (v - vh) / (vq - vh)
    return p.Lambda_H + (p.Lambda_Q - p.Lambda_H) * smoothstep(x)


def H_PT(v, p: Params):
    return p.kappa0 * v**10 * (
        1.0 + p.h2 * v**2 + p.h3s * p.a * v**3 + p.h4 * v**4
    )


def F_B(v, p: Params):
    mu_geo = M_sec(p.mu)
    M_geo = M_sec(p.M)
    return (32.0 / 5.0) * (mu_geo / M_geo)**2 * v**10 * Fhat_PP(v, p.a)


def F_T(v, p: Params):
    return F_B(v, p) * (1.0 + Lambda_of_v(v, p) * H_PT(v, p))


def dt_dv_baseline(v, p: Params):
    mu_geo = M_sec(p.mu)
    return -mu_geo * dEtilde_dv(v, p.a) / F_B(v, p)


def dt_dv_transition(v, p: Params):
    mu_geo = M_sec(p.mu)
    return -mu_geo * dEtilde_dv(v, p.a) / F_T(v, p)


# ============================================================
# simple LISA PSD
# ============================================================
def lisa_psd(f):
    L = 2.5e9
    fstar = c / (2.0 * np.pi * L)

    P_OMS = (1.5e-11)**2 * (1.0 + (2.0e-3 / np.maximum(f, 1e-12))**4)
    P_acc = (3.0e-15)**2 * (1.0 + (0.4e-3 / np.maximum(f, 1e-12))**2) * (1.0 + (f / 8.0e-3)**4)

    return (10.0 / (3.0 * L**2)) * (
        P_OMS + 4.0 * P_acc / (2.0 * np.pi * f)**4
    ) * (1.0 + 0.6 * (f / fstar)**2)


# ============================================================
# build branches
# ============================================================
def build_branches(p: Params):
    f = np.linspace(p.f_start, p.f_cut, p.nfreq)
    v = v_from_f22(f, p)

    mask = Fhat_PP(v, p.a) > 0.0
    f = f[mask]
    v = v[mask]

    dtB_dv = dt_dv_baseline(v, p)
    dtT_dv = dt_dv_transition(v, p)

    tB = cumulative_trapezoid(dtB_dv, v, initial=0.0)
    tT = cumulative_trapezoid(dtT_dv, v, initial=0.0)
    tB -= tB[-1]
    tT -= tT[-1]

    dPhiB_dv = 2.0 * np.pi * f * dtB_dv
    dPhiT_dv = 2.0 * np.pi * f * dtT_dv
    PhiB = cumulative_trapezoid(dPhiB_dv, v, initial=0.0)
    PhiT = cumulative_trapezoid(dPhiT_dv, v, initial=0.0)

    mu_geo = M_sec(p.mu)
    DL_geo = D_sec(p.DL_Gpc)

    # simple leading-order time-domain amplitude
    A_time = 2.0 * mu_geo / DL_geo * v**2

    # SPA-like frequency-domain amplitude
    df_dv = 3.0 * v**2 / (np.pi * M_sec(p.M))
    dtB_df = dtB_dv / df_dv
    dtT_df = dtT_dv / df_dv

    Atilde_B = A_time * np.sqrt(np.abs(dtB_df))
    Atilde_T = A_time * np.sqrt(np.abs(dtT_df))

    psiB = 2.0 * np.pi * f * tB - PhiB - np.pi / 4.0
    psiT = 2.0 * np.pi * f * tT - PhiT - np.pi / 4.0

    hBf = Atilde_B * np.exp(1j * psiB)
    hTf = Atilde_T * np.exp(1j * psiT)
    hRf = hBf - hTf

    return {
        "f": f,
        "v": v,
        "tB": tB,
        "tT": tT,
        "PhiB": PhiB,
        "PhiT": PhiT,
        "deltaPhi": PhiB - PhiT,
        "Lambda": Lambda_of_v(v, p),
        "FhatPP": Fhat_PP(v, p.a),
        "hBf": hBf,
        "hTf": hTf,
        "hRf": hRf,
    }


# ============================================================
# inner products and match
# ============================================================
def inner_product(h1f, h2f, f, Sn):
    return 4.0 * np.real(np.trapz(h1f * np.conj(h2f) / Sn, f))


def snr(hf, f, Sn):
    val = inner_product(hf, hf, f, Sn)
    return np.sqrt(max(val, 0.0))


def maximized_match_tc_phi(h1f, h2f, f, Sn, tc_days=0.05, ntc=1201):
    tc_grid = np.linspace(-tc_days * 86400.0, tc_days * 86400.0, ntc)

    n1 = np.sqrt(inner_product(h1f, h1f, f, Sn))
    n2 = np.sqrt(inner_product(h2f, h2f, f, Sn))
    if n1 <= 0 or n2 <= 0:
        return 0.0, 0.0, 0.0

    best_match = -np.inf
    best_tc = 0.0
    best_phase = 0.0

    for tc in tc_grid:
        shifted = h2f * np.exp(1j * 2.0 * np.pi * f * tc)
        z = 4.0 * np.trapz(h1f * np.conj(shifted) / Sn, f)
        m = np.abs(z) / (n1 * n2)
        if m > best_match:
            best_match = m
            best_tc = tc
            best_phase = -np.angle(z)

    return float(best_match), float(best_tc), float(best_phase)


# ============================================================
# calibration helpers
# ============================================================
def end_dephasing_rad(p: Params):
    return float(build_branches(p)["deltaPhi"][-1])


def calibrate_kappa0_for_target_dephasing(p: Params, target=5.0e3, low=1e2, high=1e9, niter=50):
    plo = replace(p, kappa0=low)
    phi_lo = end_dephasing_rad(plo)
    phi_hi = end_dephasing_rad(replace(p, kappa0=high))

    if not (phi_lo <= target <= phi_hi):
        return p

    a, b = low, high
    for _ in range(niter):
        mid = np.sqrt(a * b)
        pmid = replace(p, kappa0=mid)
        phi_mid = end_dephasing_rad(pmid)
        if phi_mid < target:
            a = mid
        else:
            b = mid

    return replace(p, kappa0=np.sqrt(a * b))


def calibrate_distance_for_target_rhoB(p: Params, target_rhoB=5.064):
    data = build_branches(p)
    f = data["f"]
    rho_now = snr(data["hBf"], f, lisa_psd(f))
    if rho_now <= 0:
        return p
    return replace(p, DL_Gpc=p.DL_Gpc * (rho_now / target_rhoB))


def prepare_benchmark():
    p = Params()
    p = calibrate_kappa0_for_target_dephasing(p, target=5.0e3)
    p = calibrate_distance_for_target_rhoB(p, target_rhoB=5.064)
    return p


# ============================================================
# Fisher tools
# ============================================================
def waveform_baseline_only(p: Params):
    data = build_branches(p)
    return data["f"], data["hBf"]


def finite_difference_derivative(p: Params, par_name: str, frac_eps: float):
    if par_name == "lnM":
        p_plus = replace(p, M=p.M * np.exp(frac_eps))
        p_minus = replace(p, M=p.M * np.exp(-frac_eps))
        denom = 2.0 * frac_eps
    elif par_name == "lnmu":
        p_plus = replace(p, mu=p.mu * np.exp(frac_eps))
        p_minus = replace(p, mu=p.mu * np.exp(-frac_eps))
        denom = 2.0 * frac_eps
    elif par_name == "a":
        p_plus = replace(p, a=p.a + frac_eps)
        p_minus = replace(p, a=p.a - frac_eps)
        denom = 2.0 * frac_eps
    else:
        raise ValueError("Unsupported parameter")

    f_plus, h_plus = waveform_baseline_only(p_plus)
    f_minus, h_minus = waveform_baseline_only(p_minus)

    fmin = max(f_plus.min(), f_minus.min())
    fmax = min(f_plus.max(), f_minus.max())
    f = np.linspace(fmin, fmax, min(len(f_plus), len(f_minus)))

    hpr = interp1d(f_plus, h_plus, kind="linear", fill_value=0.0, bounds_error=False)(f)
    hmr = interp1d(f_minus, h_minus, kind="linear", fill_value=0.0, bounds_error=False)(f)
    dh = (hpr - hmr) / denom
    return f, dh


def fisher_bias_analysis(p: Params):
    data = build_branches(p)
    f_full = data["f"]
    hB_full = data["hBf"]
    hT_full = data["hTf"]

    names = ["lnM", "lnmu", "a"]
    eps_map = {"lnM": p.eps_lnM, "lnmu": p.eps_lnmu, "a": p.eps_a}

    derivs = {}
    for nm in names:
        derivs[nm] = finite_difference_derivative(p, nm, eps_map[nm])

    fmin = max([f_full.min()] + [derivs[nm][0].min() for nm in names])
    fmax = min([f_full.max()] + [derivs[nm][0].max() for nm in names])
    f = np.linspace(fmin, fmax, 3000)
    Sn = lisa_psd(f)

    hB = interp1d(f_full, hB_full, kind="linear", fill_value=0.0, bounds_error=False)(f)
    hT = interp1d(f_full, hT_full, kind="linear", fill_value=0.0, bounds_error=False)(f)
    delta_h = hT - hB

    dlist = []
    for nm in names:
        fd, dwdp = derivs[nm]
        d_interp = interp1d(fd, dwdp, kind="linear", fill_value=0.0, bounds_error=False)(f)
        dlist.append(d_interp)

    npar = len(names)
    Gamma = np.zeros((npar, npar))
    bvec = np.zeros(npar)

    for i in range(npar):
        for j in range(npar):
            Gamma[i, j] = inner_product(dlist[i], dlist[j], f, Sn)
        bvec[i] = inner_product(delta_h, dlist[i], f, Sn)

    Cov = inv(Gamma)
    bias = Cov @ bvec
    sigma = np.sqrt(np.diag(Cov))

    return {
        "names": names,
        "f": f,
        "Cov": Cov,
        "Gamma": Gamma,
        "bias": bias,
        "sigma": sigma,
        "bias_over_sigma": bias / sigma,
    }


def draw_cov_ellipse(ax, mean, cov2, nsig=1.0, label=None):
    vals, vecs = eig(cov2)
    order = np.argsort(vals)[::-1]
    vals = np.real(vals[order])
    vecs = np.real(vecs[:, order])

    width = 2.0 * nsig * np.sqrt(max(vals[0], 0.0))
    height = 2.0 * nsig * np.sqrt(max(vals[1], 0.0))
    angle = np.degrees(np.arctan2(vecs[1, 0], vecs[0, 0]))

    ell = plt.matplotlib.patches.Ellipse(
        xy=mean,
        width=width,
        height=height,
        angle=angle,
        fill=False,
        lw=1.8,
        label=label
    )
    ax.add_patch(ell)
```

```python
# ============================================================
# Shared utilities for:
#   fig:refit_mismatch
#   fig:fisher_bias
#   fig:window_scan
#
# Save as, e.g., emri_transition_figures.py
# ============================================================

import os
from dataclasses import dataclass, replace

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid
from scipy.interpolate import interp1d
from numpy.linalg import inv, eig

# -------------------------------
# plotting style
# -------------------------------
plt.rcParams.update({
    "figure.dpi": 140,
    "savefig.dpi": 300,
    "font.size": 11,
    "axes.labelsize": 11,
    "axes.titlesize": 11,
    "axes.linewidth": 1.0,
    "lines.linewidth": 1.8,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 4,
    "ytick.major.size": 4,
    "xtick.minor.size": 2,
    "ytick.minor.size": 2,
    "legend.frameon": False,
})

# -------------------------------
# physical constants
# -------------------------------
G = 6.67430e-11          # gravitational constant (m^3/kg/s^2)
c = 299792458.0          # speed of light (m/s)
MSUN = 1.988409870698051e30  # solar mass (kg)
PC = 3.085677581491367e16    # parsec (m)

# Geometric units: time in seconds for a solar mass
MTSUN_SI = G * MSUN / c**3   # M_☉ in seconds (s)

@dataclass
class Params: 
    """
    Parameter container for EMRI waveform calculations.
    All values are in standard units (solar masses, Gpc, Hz, etc.).
    """
    # benchmark from the draft
    M: float = 2.0e5        # central black hole mass (M_☉)
    mu: float = 1.4         # compact object mass (M_☉)
    a: float = 0.90         # dimensionless spin of central BH

    # observational setup
    DL_Gpc: float = 1.0     # luminosity distance (Gpc)
    f_start: float = 0.0030 # starting frequency (Hz)
    f_cut: float = 0.0364   # cutoff frequency (Hz)

    # mixed-phase window
    f_h: float = 0.012      # start frequency of transition (Hz)
    f_q: float = 0.021      # end frequency of transition (Hz)

    # transition sector
    Lambda_H: float = 450.0 # high-frequency Lambda value (before transition)
    Lambda_Q: float = 5.0   # low-frequency Lambda value (after transition)

    # phenomenological kernel (tidal heating coefficients)
    kappa0: float = 2.0e6   # overall amplitude of tidal heating
    h2: float = 0.0         # 2 PN coefficient
    h3s: float = 0.0        # spin-orbit coefficient
    h4: float = 0.0         # 3 PN coefficient

    # numerics
    nfreq: int = 4000               # number of frequency points
    match_tc_days: float = 0.05     # time shift range for match maximization (days)
    ntc: int = 1201                 # number of time-shift points

    # finite differences for Fisher
    eps_lnM: float = 2.0e-3         # fractional step for ln M
    eps_lnmu: float = 2.0e-3        # fractional step for ln μ
    eps_a: float = 2.0e-4           # absolute step for spin a


# ============================================================
# unit helpers
# ============================================================
def M_sec(M_msun: float) -> float:
    """Convert mass in solar masses to geometric seconds (GM/c^3)."""
    return M_msun * MTSUN_SI

def D_sec(D_Gpc: float) -> float:
    """Convert luminosity distance in Gpc to geometric seconds (D/c)."""
    D_m = D_Gpc * 1.0e9 * PC   # meters
    return D_m / c             # seconds


# ============================================================
# Kerr circular equatorial dynamics
# ============================================================
def v_from_f22(f22, p: Params):
    """
    Convert gravitational wave frequency f22 (Hz) to the velocity parameter
    v = (M ω)^{1/3}, where ω is the orbital angular frequency.
    For the dominant (2,2) mode: ω = π f22.
    """
    return np.power(np.pi * M_sec(p.M) * f22, 1.0 / 3.0)

def f22_from_v(v, p: Params):
    """Inverse of v_from_f22."""
    return v**3 / (np.pi * M_sec(p.M))

def omega_from_v(v, p: Params):
    """Orbital angular frequency in geometric units (1/seconds)."""
    return v**3 / M_sec(p.M)

def u_of_v(v, a):
    """Dimensionless quantity u = (M ω)^{2/3} / (1 - a v^3)^{2/3}."""
    return v**2 / np.power(1.0 - a * v**3, 2.0 / 3.0)

def E_tilde(v, a):
    """Specific energy (E/μ) for a circular equatorial orbit in Kerr."""
    u = u_of_v(v, a)
    N = 1.0 - 2.0 * u + a * u**1.5
    D = 1.0 - 3.0 * u + 2.0 * a * u**1.5
    return N / np.sqrt(D)

def dEtilde_dv(v, a):
    """Derivative of specific energy with respect to v."""
    u = u_of_v(v, a)
    N = 1.0 - 2.0 * u + a * u**1.5
    D = 1.0 - 3.0 * u + 2.0 * a * u**1.5
    Np = -2.0 + 1.5 * a * np.sqrt(u)
    Dp = -3.0 + 3.0 * a * np.sqrt(u)
    dEdu = (2.0 * Np * D - N * Dp) / (2.0 * D**1.5)
    dudv = 2.0 * v / np.power(1.0 - a * v**3, 5.0 / 3.0)
    return dEdu * dudv


# ============================================================
# phenomenological transition sector
# ============================================================
def Fhat_PP(v, a):
    """
    Post-Newtonian energy flux scaled by (32/5) v^10, truncated at 2 PN.
    The full flux is (32/5) (μ/M)^2 v^10 Fhat_PP(v,a).
    Includes spin-orbit term up to 1.5 PN.
    """
    return (
        1.0
        - (1247.0 / 336.0) * v**2
        + (4.0 * np.pi - (73.0 / 12.0) * a) * v**3
        - (44711.0 / 9072.0) * v**4
    )

def smoothstep(x):
    """Cubic smoothstep function: 0 for x<=0, 1 for x>=1, smooth in between."""
    y = np.zeros_like(x)
    mid = (x >= 0.0) & (x <= 1.0)
    y[x > 1.0] = 1.0
    y[mid] = 3.0 * x[mid]**2 - 2.0 * x[mid]**3
    return y

def Lambda_of_v(v, p: Params):
    """
    Transition blending factor Λ(v). Goes from Λ_H at high frequencies (small v)
    to Λ_Q at low frequencies (large v) via a smoothstep over [vh, vq].
    """
    vh = v_from_f22(p.f_h, p)   # velocity at transition start
    vq = v_from_f22(p.f_q, p)   # velocity at transition end
    x = (v - vh) / (vq - vh)    # normalized coordinate
    return p.Lambda_H + (p.Lambda_Q - p.Lambda_H) * smoothstep(x)

def H_PT(v, p: Params):
    """Phenomenological tidal heating term, H(v)."""
    return p.kappa0 * v**10 * (
        1.0 + p.h2 * v**2 + p.h3s * p.a * v**3 + p.h4 * v**4
    )

def F_B(v, p: Params):
    """
    Energy flux for the baseline (no transition) waveform.
    Includes the mass ratio factor (μ/M)^2.
    """
    mu_geo = M_sec(p.mu)
    M_geo = M_sec(p.M)
    return (32.0 / 5.0) * (mu_geo / M_geo)**2 * v**10 * Fhat_PP(v, p.a)

def F_T(v, p: Params):
    """
    Energy flux for the transition waveform, which includes tidal heating
    via the factor (1 + Λ H).
    """
    return F_B(v, p) * (1.0 + Lambda_of_v(v, p) * H_PT(v, p))

def dt_dv_baseline(v, p: Params):
    """
    dt/dv for the baseline waveform (no tidal heating).
    dt/dv = - μ (dE/dv) / F_B.
    """
    mu_geo = M_sec(p.mu)
    return -mu_geo * dEtilde_dv(v, p.a) / F_B(v, p)

def dt_dv_transition(v, p: Params):
    """
    dt/dv for the transition waveform (with tidal heating).
    """
    mu_geo = M_sec(p.mu)
    return -mu_geo * dEtilde_dv(v, p.a) / F_T(v, p)


# ============================================================
# simple LISA PSD
# ============================================================
def lisa_psd(f):
    """
    One-sided power spectral density of LISA noise, including optical metrology
    and acceleration noise, with simplified transfer functions.
    """
    L = 2.5e9                           # arm length (m)
    fstar = c / (2.0 * np.pi * L)       # characteristic frequency of the instrument

    # Optical metrology noise (modified with 1/f^4 tail)
    P_OMS = (1.5e-11)**2 * (1.0 + (2.0e-3 / np.maximum(f, 1e-12))**4)
    # Acceleration noise (including low-frequency 1/f^2 and high-frequency plateau)
    P_acc = (3.0e-15)**2 * (1.0 + (0.4e-3 / np.maximum(f, 1e-12))**2) * (1.0 + (f / 8.0e-3)**4)

    # Combine to form the full PSD (one-sided, for Michelson channels)
    return (10.0 / (3.0 * L**2)) * (
        P_OMS + 4.0 * P_acc / (2.0 * np.pi * f)**4
    ) * (1.0 + 0.6 * (f / fstar)**2)


# ============================================================
# build branches
# ============================================================
def build_branches(p: Params):
    """
    Construct the frequency-domain waveforms for baseline and transition,
    along with auxiliary quantities like time, phase, and amplitude.

    Returns a dictionary containing:
        - f: frequency grid [Hz]
        - v: velocity grid
        - tB, tT: time of coalescence (relative) for baseline and transition
        - PhiB, PhiT: orbital phase (not gravitational phase)
        - deltaPhi: difference in orbital phase (baseline - transition)
        - Lambda: blending factor
        - FhatPP: PN flux (without mass ratio)
        - hBf, hTf, hRf: complex frequency-domain strains (baseline, transition, residual)
    """
    f = np.linspace(p.f_start, p.f_cut, p.nfreq)
    v = v_from_f22(f, p)

    # Ensure flux is positive (remove points where Fhat_PP becomes negative)
    mask = Fhat_PP(v, p.a) > 0.0
    f = f[mask]
    v = v[mask]

    # Compute dt/dv for both branches
    dtB_dv = dt_dv_baseline(v, p)
    dtT_dv = dt_dv_transition(v, p)

    # Integrate to get time to merger (t = 0 at merger, so we shift later)
    tB = cumulative_trapezoid(dtB_dv, v, initial=0.0)
    tT = cumulative_trapezoid(dtT_dv, v, initial=0.0)
    tB -= tB[-1]    # set t=0 at merger
    tT -= tT[-1]

    # Phase derivatives: dΦ/dv = 2π f dt/dv (since dΦ = 2π f dt)
    dPhiB_dv = 2.0 * np.pi * f * dtB_dv
    dPhiT_dv = 2.0 * np.pi * f * dtT_dv
    PhiB = cumulative_trapezoid(dPhiB_dv, v, initial=0.0)
    PhiT = cumulative_trapezoid(dPhiT_dv, v, initial=0.0)

    # Newtonian amplitude in time domain (v^2 scaling)
    mu_geo = M_sec(p.mu)
    DL_geo = D_sec(p.DL_Gpc)
    A_time = 2.0 * mu_geo / DL_geo * v**2

    # Convert to frequency domain using stationary phase approximation
    # Amplitude: A_f = A_time * sqrt(|dt/df|)
    df_dv = 3.0 * v**2 / (np.pi * M_sec(p.M))   # df/dv from f = v^3/(π M)
    dtB_df = dtB_dv / df_dv
    dtT_df = dtT_dv / df_dv
    Atilde_B = A_time * np.sqrt(np.abs(dtB_df))
    Atilde_T = A_time * np.sqrt(np.abs(dtT_df))

    # Phase in frequency domain: ψ = 2π f t - Φ - π/4
    psiB = 2.0 * np.pi * f * tB - PhiB - np.pi / 4.0
    psiT = 2.0 * np.pi * f * tT - PhiT - np.pi / 4.0

    # Frequency-domain strain (complex)
    hBf = Atilde_B * np.exp(1j * psiB)
    hTf = Atilde_T * np.exp(1j * psiT)
    hRf = hBf - hTf   # residual

    return {
        "f": f,
        "v": v,
        "tB": tB,
        "tT": tT,
        "PhiB": PhiB,
        "PhiT": PhiT,
        "deltaPhi": PhiB - PhiT,   # orbital phase difference
        "Lambda": Lambda_of_v(v, p),
        "FhatPP": Fhat_PP(v, p.a),
        "hBf": hBf,
        "hTf": hTf,
        "hRf": hRf,
    }


# ============================================================
# inner products and match
# ============================================================
def inner_product(h1f, h2f, f, Sn):
    """Noise-weighted inner product: <h1|h2> = 4 Re ∫ h1 h2* / Sn df."""
    return 4.0 * np.real(np.trapz(h1f * np.conj(h2f) / Sn, f))

def snr(hf, f, Sn):
    """Signal-to-noise ratio: sqrt(<h|h>)."""
    val = inner_product(hf, hf, f, Sn)
    return np.sqrt(max(val, 0.0))

def maximized_match_tc_phi(h1f, h2f, f, Sn, tc_days=0.05, ntc=1201):
    """
    Compute the match (normalized inner product) between two frequency-domain
    waveforms, maximizing over a time shift (tc) and a constant phase shift.
    The time shift is searched over [-tc_days, tc_days] days.
    """
    # Create time shift grid (seconds)
    tc_grid = np.linspace(-tc_days * 86400.0, tc_days * 86400.0, ntc)

    # Norms of the two waveforms
    n1 = np.sqrt(inner_product(h1f, h1f, f, Sn))
    n2 = np.sqrt(inner_product(h2f, h2f, f, Sn))
    if n1 <= 0 or n2 <= 0:
        return 0.0, 0.0, 0.0

    best_match = -np.inf
    best_tc = 0.0
    best_phase = 0.0

    for tc in tc_grid:
        # Apply time shift: h2(f) -> h2(f) exp(2π i f tc)
        shifted = h2f * np.exp(1j * 2.0 * np.pi * f * tc)
        # Overlap without phase optimization: ∫ h1* conj(shifted) / Sn df
        z = 4.0 * np.trapz(h1f * np.conj(shifted) / Sn, f)
        # Match after maximizing over phase: |z|/(n1 n2)
        m = np.abs(z) / (n1 * n2)
        if m > best_match:
            best_match = m
            best_tc = tc
            best_phase = -np.angle(z)   # optimal constant phase shift

    return float(best_match), float(best_tc), float(best_phase)


# ============================================================
# calibration helpers
# ============================================================
def end_dephasing_rad(p: Params):
    """Compute the total accumulated orbital phase dephasing at the end of the inspiral."""
    return float(build_branches(p)["deltaPhi"][-1])

def calibrate_kappa0_for_target_dephasing(p: Params, target=5.0e3, low=1e2, high=1e9, niter=50):
    """
    Tune the tidal heating amplitude kappa0 so that the total dephasing
    at the end of the inspiral equals a target value (default 5000 rad).
    Uses a logarithmic bisection search.
    """
    plo = replace(p, kappa0=low)
    phi_lo = end_dephasing_rad(plo)
    phi_hi = end_dephasing_rad(replace(p, kappa0=high))

    if not (phi_lo <= target <= phi_hi):
        return p   # target out of range, return unchanged

    a, b = low, high
    for _ in range(niter):
        mid = np.sqrt(a * b)
        pmid = replace(p, kappa0=mid)
        phi_mid = end_dephasing_rad(pmid)
        if phi_mid < target:
            a = mid
        else:
            b = mid
    return replace(p, kappa0=np.sqrt(a * b))

def calibrate_distance_for_target_rhoB(p: Params, target_rhoB=5.064):
    """
    Scale the luminosity distance so that the baseline SNR matches a given target.
    SNR scales as 1/DL, so we multiply DL by (current SNR / target SNR).
    """
    data = build_branches(p)
    f = data["f"]
    rho_now = snr(data["hBf"], f, lisa_psd(f))
    if rho_now <= 0:
        return p
    return replace(p, DL_Gpc=p.DL_Gpc * (rho_now / target_rhoB))

def prepare_benchmark():
    """
    Construct a benchmark parameter set with calibrated kappa0 to achieve
    ~5000 rad dephasing and a baseline SNR of ~5.064.
    """
    p = Params()
    p = calibrate_kappa0_for_target_dephasing(p, target=5.0e3)
    p = calibrate_distance_for_target_rhoB(p, target_rhoB=5.064)
    return p


# ============================================================
# Fisher tools
# ============================================================
def waveform_baseline_only(p: Params):
    """
    Return the frequency grid and baseline waveform (no transition) only.
    Used for Fisher matrix calculations.
    """
    data = build_branches(p)
    return data["f"], data["hBf"]

def finite_difference_derivative(p: Params, par_name: str, frac_eps: float):
    """
    Compute the partial derivative ∂h/∂λ via symmetric finite difference.
    Supports parameters: lnM, lnmu, a.
    Returns (frequency_grid, derivative).
    """
    if par_name == "lnM":
        p_plus = replace(p, M=p.M * np.exp(frac_eps))
        p_minus = replace(p, M=p.M * np.exp(-frac_eps))
        denom = 2.0 * frac_eps
    elif par_name == "lnmu":
        p_plus = replace(p, mu=p.mu * np.exp(frac_eps))
        p_minus = replace(p, mu=p.mu * np.exp(-frac_eps))
        denom = 2.0 * frac_eps
    elif par_name == "a":
        p_plus = replace(p, a=p.a + frac_eps)
        p_minus = replace(p, a=p.a - frac_eps)
        denom = 2.0 * frac_eps
    else:
        raise ValueError("Unsupported parameter")

    # Get waveforms for both perturbed parameters
    f_plus, h_plus = waveform_baseline_only(p_plus)
    f_minus, h_minus = waveform_baseline_only(p_minus)

    # Common frequency range
    fmin = max(f_plus.min(), f_minus.min())
    fmax = min(f_plus.max(), f_minus.max())
    f = np.linspace(fmin, fmax, min(len(f_plus), len(f_minus)))

    # Interpolate to common grid
    hpr = interp1d(f_plus, h_plus, kind="linear", fill_value=0.0, bounds_error=False)(f)
    hmr = interp1d(f_minus, h_minus, kind="linear", fill_value=0.0, bounds_error=False)(f)
    dh = (hpr - hmr) / denom
    return f, dh

def fisher_bias_analysis(p: Params):
    """
    Compute the Fisher matrix and the expected bias (in the linear approximation)
    when the true waveform is the transition waveform and the template family
    is the baseline waveform (i.e., neglecting tidal heating).
    Returns a dictionary with:
        - names: parameter names
        - f: frequency grid used
        - Cov: covariance matrix (inverse Fisher)
        - Gamma: Fisher matrix
        - bias: bias vector (δθ)
        - sigma: 1σ errors from Fisher
        - bias_over_sigma: bias normalized by standard deviation
    """
    data = build_branches(p)
    f_full = data["f"]
    hB_full = data["hBf"]          # baseline waveform
    hT_full = data["hTf"]          # transition (true) waveform

    # Parameter names and their finite-difference step sizes
    names = ["lnM", "lnmu", "a"]
    eps_map = {"lnM": p.eps_lnM, "lnmu": p.eps_lnmu, "a": p.eps_a}

    # Compute derivatives for each parameter
    derivs = {}
    for nm in names:
        derivs[nm] = finite_difference_derivative(p, nm, eps_map[nm])

    # Common frequency grid for the Fisher calculation
    fmin = max([f_full.min()] + [derivs[nm][0].min() for nm in names])
    fmax = min([f_full.max()] + [derivs[nm][0].max() for nm in names])
    f = np.linspace(fmin, fmax, 3000)
    Sn = lisa_psd(f)

    # Interpolate waveforms onto the common grid
    hB = interp1d(f_full, hB_full, kind="linear", fill_value=0.0, bounds_error=False)(f)
    hT = interp1d(f_full, hT_full, kind="linear", fill_value=0.0, bounds_error=False)(f)
    delta_h = hT - hB   # difference between true and template

    # Interpolate derivatives onto the common grid
    dlist = []
    for nm in names:
        fd, dwdp = derivs[nm]
        d_interp = interp1d(fd, dwdp, kind="linear", fill_value=0.0, bounds_error=False)(f)
        dlist.append(d_interp)

    npar = len(names)
    Gamma = np.zeros((npar, npar))   # Fisher matrix
    bvec = np.zeros(npar)            # bias vector numerator

    for i in range(npar):
        for j in range(npar):
            Gamma[i, j] = inner_product(dlist[i], dlist[j], f, Sn)
        bvec[i] = inner_product(delta_h, dlist[i], f, Sn)

    Cov = inv(Gamma)                 # covariance matrix
    bias = Cov @ bvec                # bias estimate: δθ = Γ^{-1} b
    sigma = np.sqrt(np.diag(Cov))    # standard deviations

    return {
        "names": names,
        "f": f,
        "Cov": Cov,
        "Gamma": Gamma,
        "bias": bias,
        "sigma": sigma,
        "bias_over_sigma": bias / sigma,
    }

def draw_cov_ellipse(ax, mean, cov2, nsig=1.0, label=None):
    """
    Draw a covariance ellipse on an axis.
    Parameters:
        ax: matplotlib axes
        mean: (x, y) center
        cov2: 2x2 covariance matrix
        nsig: number of sigma (size of ellipse)
        label: optional label for the ellipse
    """
    # Compute eigenvalues and eigenvectors
    vals, vecs = eig(cov2)
    order = np.argsort(vals)[::-1]
    vals = np.real(vals[order])
    vecs = np.real(vecs[:, order])

    width = 2.0 * nsig * np.sqrt(max(vals[0], 0.0))
    height = 2.0 * nsig * np.sqrt(max(vals[1], 0.0))
    angle = np.degrees(np.arctan2(vecs[1, 0], vecs[0, 0]))

    ell = plt.matplotlib.patches.Ellipse(
        xy=mean,
        width=width,
        height=height,
        angle=angle,
        fill=False,
        lw=1.8,
        label=label
    )
    ax.add_patch(ell)
```

```python
# ============================================================
# fig:refit_mismatch
# ============================================================
def fig_refit_mismatch(
    p: Params,
    outfile="fig_refit_mismatch.pdf",
    plane="M-a",
    ngrid=41
):
    """
    Generate a 2D mismatch map showing how well the baseline (no transition)
    template family can fit the true transition waveform, as a function of
    two parameters (M-a or M-mu).

    The true waveform is the transition model with a fixed parameter set.
    For each point on a grid of two parameters, we compute the mismatch
    (1 - optimized match) between the true waveform and a baseline template
    evaluated at that point. The best-fitting point (lowest mismatch) is also
    located and reported.

    Parameters
    ----------
    p : Params
        Parameter object containing the true parameters and all model settings.
    outfile : str, optional
        Output filename for the saved figure.
    plane : str, optional
        Which parameter plane to scan:
            "M-a"  -> masses (M) and spin (a)
            "M-mu" -> masses (M) and secondary mass (mu)
    ngrid : int, optional
        Number of grid points in each dimension.

    Returns
    -------
    best : dict
        Contains 'match' (maximum match found), 'x', 'y' (parameter values),
        'tc', 'phi' (optimized time and phase shifts).
    mismatch : ndarray
        2D array of mismatch values (1 - match) on the grid.
    """
    # Build the true transition waveform (includes tidal heating)
    data_true = build_branches(p)
    f_true = data_true["f"]
    h_true = data_true["hTf"]

    # Set up parameter grids for the two dimensions
    if plane == "M-a":
        # M range: ±4% around true value
        xvals = np.linspace(0.96 * p.M, 1.04 * p.M, ngrid)
        # a range: ±0.08, but clamped to reasonable bounds [0.5, 0.999)
        yvals = np.linspace(max(0.5, p.a - 0.08), min(0.999, p.a + 0.08), ngrid)
        xlabel = r"$M\,[10^5 M_\odot]$"
        ylabel = r"$\hat a$"
    elif plane == "M-mu":
        xvals = np.linspace(0.96 * p.M, 1.04 * p.M, ngrid)
        yvals = np.linspace(0.85 * p.mu, 1.15 * p.mu, ngrid)
        xlabel = r"$M\,[10^5 M_\odot]$"
        ylabel = r"$\mu\,[M_\odot]$"
    else:
        raise ValueError("plane must be 'M-a' or 'M-mu'")

    mismatch = np.full((len(yvals), len(xvals)), np.nan)
    best = {"match": -np.inf, "x": None, "y": None, "tc": None, "phi": None}

    # Loop over the 2D grid
    for iy, y in enumerate(yvals):
        for ix, x in enumerate(xvals):
            # Build a new parameter object with the current trial values
            if plane == "M-a":
                pb = replace(p, M=float(x), a=float(y))
            else:
                pb = replace(p, M=float(x), mu=float(y))

            # Compute the baseline waveform for the trial parameters
            model = build_branches(pb)

            # Determine the overlapping frequency range between true and trial waveforms
            fmin = max(f_true.min(), model["f"].min())
            fmax = min(f_true.max(), model["f"].max())
            if fmax <= fmin:
                continue

            # Interpolate both waveforms onto a common high‑resolution frequency grid
            f = np.linspace(fmin, fmax, 2500)
            Sn = lisa_psd(f)

            hT = interp1d(f_true, h_true, kind="linear", fill_value=0.0, bounds_error=False)(f)
            hB = interp1d(model["f"], model["hBf"], kind="linear", fill_value=0.0, bounds_error=False)(f)

            # Compute the match optimized over time and phase shifts
            match, tc_best, phi_best = maximized_match_tc_phi(
                hT, hB, f, Sn, tc_days=p.match_tc_days, ntc=p.ntc
            )
            mm = 1.0 - match
            mismatch[iy, ix] = mm

            # Track the best match found so far
            if match > best["match"]:
                best = {"match": match, "x": x, "y": y, "tc": tc_best, "phi": phi_best}

    # Plot the mismatch map
    fig, ax = plt.subplots(figsize=(6.2, 5.1))

    # Extent for imshow: [xmin, xmax, ymin, ymax] (with x in units of 10^5 M_sun)
    extent = [
        xvals[0] / 1e5, xvals[-1] / 1e5,
        yvals[0], yvals[-1]
    ]

    im = ax.imshow(
        mismatch,
        origin="lower",
        aspect="auto",
        extent=extent
    )
    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label(r"Mismatch  $1-\mathrm{Match}$")

    # Mark the true parameters (the transition benchmark)
    if plane == "M-a":
        ax.scatter([p.M / 1e5], [p.a], marker="x", s=70, label="transition benchmark")
    else:
        ax.scatter([p.M / 1e5], [p.mu], marker="x", s=70, label="transition benchmark")

    # Mark the best‑fitting baseline template point
    ax.scatter(
        [best["x"] / 1e5],
        [best["y"]],
        marker="o",
        s=50,
        facecolors="none",
        label="best baseline refit"
    )

    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    ax.legend(loc="best")

    # Add a text box with the minimal mismatch and the optimal time shift
    txt = (
        rf"$\mathcal{{M}}_{{\min}} = {1.0 - best['match']:.3e}$" + "\n" +
        rf"$t_c^* = {best['tc']:.2f}\,\mathrm{{s}}$"
    )
    ax.text(
        0.03, 0.97, txt,
        transform=ax.transAxes,
        va="top",
        ha="left",
        bbox=dict(boxstyle="round", facecolor="white", alpha=0.8)
    )

    fig.tight_layout()
    fig.savefig(outfile, bbox_inches="tight")
    plt.close(fig)

    return best, mismatch
```

```python
# ============================================================
# fig:fisher_bias
# ============================================================
def fig_fisher_bias(
    p: Params,
    outfile="fig_fisher_bias.pdf"
):
    """
    Generate a three-panel figure showing the Fisher matrix bias analysis.
    The figure includes:
        (a) Bar plot of the absolute bias-to-sigma ratio for each parameter,
            highlighting whether the bias exceeds the 1σ level.
        (b) 1σ and 2σ confidence ellipses in the (ln M, a) plane, with an arrow
            indicating the systematic bias vector.
        (c) Similar ellipses in the (ln M, ln μ) plane.

    Parameters
    ----------
    p : Params
        Parameter object containing the true parameters and all model settings.
    outfile : str, optional
        Output filename for the saved figure.

    Returns
    -------
    res : dict
        The full result dictionary from fisher_bias_analysis(), containing
        the Fisher matrix, covariance, bias, and other diagnostics.
    """
    # Compute Fisher matrix, bias, and covariance using the baseline template family
    res = fisher_bias_analysis(p)

    # Extract relevant quantities
    names = res["names"]               # parameter names: ["lnM", "lnmu", "a"]
    bias = res["bias"]                 # systematic bias vector
    sigma = res["sigma"]               # 1σ statistical uncertainties
    bias_over_sigma = np.abs(res["bias_over_sigma"])  # absolute bias-to-sigma ratio
    Cov = res["Cov"]                   # covariance matrix (inverse Fisher)

    # Pretty parameter labels for plotting
    pretty = [r"$\ln M$", r"$\ln \mu$", r"$\hat a$"]

    # Create figure with three subplots
    fig = plt.figure(figsize=(10.6, 4.2))

    # --------------------------------
    # panel (a): bias significance
    # --------------------------------
    ax1 = fig.add_subplot(1, 3, 1)
    x = np.arange(len(names))
    ax1.bar(x, bias_over_sigma)
    # Add horizontal line at 1σ threshold
    ax1.axhline(1.0, ls="--", lw=1.2, color="gray")
    ax1.set_xticks(x)
    ax1.set_xticklabels(pretty)
    ax1.set_ylabel(r"$|\Delta \theta_{\rm sys}|/\sigma$")
    ax1.set_title(r"(a)", loc="left")

    # --------------------------------
    # panel (b): (ln M, a) ellipse
    # --------------------------------
    ax2 = fig.add_subplot(1, 3, 2)
    idx_M, idx_a = 0, 2                     # indices for ln M and a
    cov2 = Cov[np.ix_([idx_M, idx_a], [idx_M, idx_a])]   # 2×2 submatrix
    # Draw 1σ and 2σ ellipses centered at (0,0) (bias is shown as a shift)
    draw_cov_ellipse(ax2, np.array([0.0, 0.0]), cov2, nsig=1.0, label=r"$1\sigma$")
    draw_cov_ellipse(ax2, np.array([0.0, 0.0]), cov2, nsig=2.0, label=r"$2\sigma$")
    # Arrow representing the systematic bias vector (from true parameters)
    ax2.arrow(
        0.0, 0.0, bias[idx_M], bias[idx_a],
        width=0.0,                       # default arrow width (matplotlib)
        length_includes_head=True,       # head is part of the arrow length
        color="red"
    )
    ax2.set_xlabel(r"$\Delta \ln M$")
    ax2.set_ylabel(r"$\Delta \hat a$")
    ax2.set_title(r"(b)", loc="left")
    ax2.legend(loc="best")

    # --------------------------------
    # panel (c): (ln M, ln mu) ellipse
    # --------------------------------
    ax3 = fig.add_subplot(1, 3, 3)
    idx_M, idx_mu = 0, 1                   # indices for ln M and ln μ
    cov2b = Cov[np.ix_([idx_M, idx_mu], [idx_M, idx_mu])]   # 2×2 submatrix
    draw_cov_ellipse(ax3, np.array([0.0, 0.0]), cov2b, nsig=1.0, label=r"$1\sigma$")
    draw_cov_ellipse(ax3, np.array([0.0, 0.0]), cov2b, nsig=2.0, label=r"$2\sigma$")
    ax3.arrow(
        0.0, 0.0, bias[idx_M], bias[idx_mu],
        width=0.0,
        length_includes_head=True,
        color="red"
    )
    ax3.set_xlabel(r"$\Delta \ln M$")
    ax3.set_ylabel(r"$\Delta \ln \mu$")
    ax3.set_title(r"(c)", loc="left")
    ax3.legend(loc="best")

    # Adjust layout and save
    fig.tight_layout()
    fig.savefig(outfile, bbox_inches="tight")
    plt.close(fig)

    return res
```

```python
# ============================================================
# fig:window_scan
# ============================================================
def fig_window_scan(
    p: Params,
    outfile="fig_window_scan.pdf",
    ncenter=34,
    nwidth=30
):
    """
    Generate a three-panel figure scanning the transition window parameters.
    The window is defined by its center frequency f_c and width w, with
    transition start f_h = f_c - w/2 and end f_q = f_c + w/2.

    For each (center, width) combination, we:
        - Compute the transition waveform with that window,
        - Evaluate the mismatch (1 - match) between the baseline and transition,
        - Compute the residual SNR (SNR of the difference hT - hB),
        - Record the total orbital dephasing at the end of the inspiral.

    The three panels show:
        (1) Mismatch (1 - match) as a function of window parameters,
        (2) Residual SNR,
        (3) Total orbital dephasing.

    Parameters
    ----------
    p : Params
        Base parameter object (the window parameters f_h and f_q will be overridden).
    outfile : str, optional
        Output filename for the saved figure.
    ncenter : int, optional
        Number of points in the window center scan (f_c).
    nwidth : int, optional
        Number of points in the window width scan (w).

    Returns
    -------
    dict : Contains the grid data:
        - centers: array of center frequencies [Hz]
        - widths: array of window widths [Hz]
        - mismatch: 2D array of mismatch values
        - rhoR: 2D array of residual SNR values
        - dphi: 2D array of total dephasing values (radians)
    """
    # Define the scan grids
    centers = np.linspace(0.011, 0.024, ncenter)      # window center frequencies [Hz]
    widths = np.linspace(0.003, 0.012, nwidth)        # window widths [Hz]

    # Initialize output arrays
    MM = np.full((len(widths), len(centers)), np.nan)  # mismatch (1 - match)
    RR = np.full((len(widths), len(centers)), np.nan)  # residual SNR
    DP = np.full((len(widths), len(centers)), np.nan)  # total orbital dephasing

    # Loop over all combinations
    for iw, w in enumerate(widths):
        for ic, fc in enumerate(centers):
            f_h = fc - 0.5 * w
            f_q = fc + 0.5 * w

            # Skip invalid windows (outside the overall frequency range or degenerate)
            if f_h <= p.f_start or f_q >= p.f_cut or f_q <= f_h:
                continue

            # Create a new parameter object with the current window
            ps = replace(p, f_h=float(f_h), f_q=float(f_q))

            # Build the waveforms (baseline and transition) for this window
            data = build_branches(ps)
            f = data["f"]
            Sn = lisa_psd(f)

            # Compute residual SNR: SNR of hT - hB
            rhoR = snr(data["hRf"], f, Sn)

            # Compute mismatch between transition (true) and baseline (template)
            # with optimization over time and phase shifts
            match, _, _ = maximized_match_tc_phi(
                data["hTf"], data["hBf"], f, Sn,
                tc_days=p.match_tc_days,
                ntc=max(401, p.ntc // 2)   # use at least 401 points for time shift
            )

            # Store results
            MM[iw, ic] = 1.0 - match         # mismatch
            RR[iw, ic] = rhoR                # residual SNR
            DP[iw, ic] = data["deltaPhi"][-1] # total dephasing (at the end)

    # Create the figure with three subplots
    fig = plt.figure(figsize=(12.6, 4.0))

    # Panel (a): Mismatch
    ax1 = fig.add_subplot(1, 3, 1)
    im1 = ax1.imshow(
        MM,
        origin="lower",
        aspect="auto",
        extent=[centers[0], centers[-1], widths[0], widths[-1]]
    )
    cbar1 = plt.colorbar(im1, ax=ax1)
    cbar1.set_label(r"Mismatch  $\mathcal{M}$")
    ax1.set_xlabel(r"window center $f_c$ [Hz]")
    ax1.set_ylabel(r"window width $w$ [Hz]")
    ax1.set_title(r"(a)", loc="left")

    # Panel (b): Residual SNR
    ax2 = fig.add_subplot(1, 3, 2)
    im2 = ax2.imshow(
        RR,
        origin="lower",
        aspect="auto",
        extent=[centers[0], centers[-1], widths[0], widths[-1]]
    )
    cbar2 = plt.colorbar(im2, ax=ax2)
    cbar2.set_label(r"Residual SNR  $\rho_{\rm R}$")
    ax2.set_xlabel(r"window center $f_c$ [Hz]")
    ax2.set_ylabel(r"window width $w$ [Hz]")
    ax2.set_title(r"(b)", loc="left")

    # Panel (c): Total orbital dephasing
    ax3 = fig.add_subplot(1, 3, 3)
    im3 = ax3.imshow(
        DP,
        origin="lower",
        aspect="auto",
        extent=[centers[0], centers[-1], widths[0], widths[-1]]
    )
    cbar3 = plt.colorbar(im3, ax=ax3)
    cbar3.set_label(r"$\Delta\Phi_{22}^{\rm SF}$ [rad]")
    ax3.set_xlabel(r"window center $f_c$ [Hz]")
    ax3.set_ylabel(r"window width $w$ [Hz]")
    ax3.set_title(r"(c)", loc="left")

    fig.tight_layout()
    fig.savefig(outfile, bbox_inches="tight")
    plt.close(fig)

    # Return the grid data for further analysis
    return {"centers": centers, "widths": widths, "mismatch": MM, "rhoR": RR, "dphi": DP}
```
