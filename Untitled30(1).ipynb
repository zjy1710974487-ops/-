```python
# ============================================================
# Shared utilities for:
#   fig:refit_mismatch
#   fig:fisher_bias
#   fig:window_scan
#
# Save as, e.g., emri_transition_figures.py
# ============================================================

import os
from dataclasses import dataclass, replace

import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import cumulative_trapezoid
from scipy.interpolate import interp1d
from numpy.linalg import inv, eig

# -------------------------------
# plotting style
# -------------------------------
plt.rcParams.update({
    "figure.dpi": 140,
    "savefig.dpi": 300,
    "font.size": 11,
    "axes.labelsize": 11,
    "axes.titlesize": 11,
    "axes.linewidth": 1.0,
    "lines.linewidth": 1.8,
    "xtick.direction": "in",
    "ytick.direction": "in",
    "xtick.major.size": 4,
    "ytick.major.size": 4,
    "xtick.minor.size": 2,
    "ytick.minor.size": 2,
    "legend.frameon": False,
})

# -------------------------------
# physical constants
# -------------------------------
G = 6.67430e-11
c = 299792458.0
MSUN = 1.988409870698051e30
PC = 3.085677581491367e16

MTSUN_SI = G * MSUN / c**3


@dataclass
class Params:
    # benchmark from the draft
    M: float = 2.0e5
    mu: float = 1.4
    a: float = 0.90

    # observational setup
    DL_Gpc: float = 1.0
    f_start: float = 0.0030
    f_cut: float = 0.0364

    # mixed-phase window
    f_h: float = 0.012
    f_q: float = 0.021

    # transition sector
    Lambda_H: float = 450.0
    Lambda_Q: float = 5.0

    # phenomenological kernel
    # these are tunable because your manuscript fixes the structure
    # but not a unique final coefficient set
    kappa0: float = 2.0e6
    h2: float = 0.0
    h3s: float = 0.0
    h4: float = 0.0

    # numerics
    nfreq: int = 4000
    match_tc_days: float = 0.05
    ntc: int = 1201

    # finite differences for Fisher
    eps_lnM: float = 2.0e-3
    eps_lnmu: float = 2.0e-3
    eps_a: float = 2.0e-4


# ============================================================
# unit helpers
# ============================================================
def M_sec(M_msun: float) -> float:
    return M_msun * MTSUN_SI


def D_sec(D_Gpc: float) -> float:
    D_m = D_Gpc * 1.0e9 * PC
    return D_m / c


# ============================================================
# Kerr circular equatorial dynamics
# ============================================================
def v_from_f22(f22, p: Params):
    return np.power(np.pi * M_sec(p.M) * f22, 1.0 / 3.0)


def f22_from_v(v, p: Params):
    return v**3 / (np.pi * M_sec(p.M))


def omega_from_v(v, p: Params):
    return v**3 / M_sec(p.M)


def u_of_v(v, a):
    return v**2 / np.power(1.0 - a * v**3, 2.0 / 3.0)


def E_tilde(v, a):
    u = u_of_v(v, a)
    N = 1.0 - 2.0 * u + a * u**1.5
    D = 1.0 - 3.0 * u + 2.0 * a * u**1.5
    return N / np.sqrt(D)


def dEtilde_dv(v, a):
    u = u_of_v(v, a)
    N = 1.0 - 2.0 * u + a * u**1.5
    D = 1.0 - 3.0 * u + 2.0 * a * u**1.5
    Np = -2.0 + 1.5 * a * np.sqrt(u)
    Dp = -3.0 + 3.0 * a * np.sqrt(u)
    dEdu = (2.0 * Np * D - N * Dp) / (2.0 * D**1.5)
    dudv = 2.0 * v / np.power(1.0 - a * v**3, 5.0 / 3.0)
    return dEdu * dudv


# ============================================================
# phenomenological transition sector
# ============================================================
def Fhat_PP(v, a):
    return (
        1.0
        - (1247.0 / 336.0) * v**2
        + (4.0 * np.pi - (73.0 / 12.0) * a) * v**3
        - (44711.0 / 9072.0) * v**4
    )


def smoothstep(x):
    y = np.zeros_like(x)
    mid = (x >= 0.0) & (x <= 1.0)
    y[x > 1.0] = 1.0
    y[mid] = 3.0 * x[mid]**2 - 2.0 * x[mid]**3
    return y


def Lambda_of_v(v, p: Params):
    vh = v_from_f22(p.f_h, p)
    vq = v_from_f22(p.f_q, p)
    x = (v - vh) / (vq - vh)
    return p.Lambda_H + (p.Lambda_Q - p.Lambda_H) * smoothstep(x)


def H_PT(v, p: Params):
    return p.kappa0 * v**10 * (
        1.0 + p.h2 * v**2 + p.h3s * p.a * v**3 + p.h4 * v**4
    )


def F_B(v, p: Params):
    mu_geo = M_sec(p.mu)
    M_geo = M_sec(p.M)
    return (32.0 / 5.0) * (mu_geo / M_geo)**2 * v**10 * Fhat_PP(v, p.a)


def F_T(v, p: Params):
    return F_B(v, p) * (1.0 + Lambda_of_v(v, p) * H_PT(v, p))


def dt_dv_baseline(v, p: Params):
    mu_geo = M_sec(p.mu)
    return -mu_geo * dEtilde_dv(v, p.a) / F_B(v, p)


def dt_dv_transition(v, p: Params):
    mu_geo = M_sec(p.mu)
    return -mu_geo * dEtilde_dv(v, p.a) / F_T(v, p)


# ============================================================
# simple LISA PSD
# ============================================================
def lisa_psd(f):
    L = 2.5e9
    fstar = c / (2.0 * np.pi * L)

    P_OMS = (1.5e-11)**2 * (1.0 + (2.0e-3 / np.maximum(f, 1e-12))**4)
    P_acc = (3.0e-15)**2 * (1.0 + (0.4e-3 / np.maximum(f, 1e-12))**2) * (1.0 + (f / 8.0e-3)**4)

    return (10.0 / (3.0 * L**2)) * (
        P_OMS + 4.0 * P_acc / (2.0 * np.pi * f)**4
    ) * (1.0 + 0.6 * (f / fstar)**2)


# ============================================================
# build branches
# ============================================================
def build_branches(p: Params):
    f = np.linspace(p.f_start, p.f_cut, p.nfreq)
    v = v_from_f22(f, p)

    mask = Fhat_PP(v, p.a) > 0.0
    f = f[mask]
    v = v[mask]

    dtB_dv = dt_dv_baseline(v, p)
    dtT_dv = dt_dv_transition(v, p)

    tB = cumulative_trapezoid(dtB_dv, v, initial=0.0)
    tT = cumulative_trapezoid(dtT_dv, v, initial=0.0)
    tB -= tB[-1]
    tT -= tT[-1]

    dPhiB_dv = 2.0 * np.pi * f * dtB_dv
    dPhiT_dv = 2.0 * np.pi * f * dtT_dv
    PhiB = cumulative_trapezoid(dPhiB_dv, v, initial=0.0)
    PhiT = cumulative_trapezoid(dPhiT_dv, v, initial=0.0)

    mu_geo = M_sec(p.mu)
    DL_geo = D_sec(p.DL_Gpc)

    # simple leading-order time-domain amplitude
    A_time = 2.0 * mu_geo / DL_geo * v**2

    # SPA-like frequency-domain amplitude
    df_dv = 3.0 * v**2 / (np.pi * M_sec(p.M))
    dtB_df = dtB_dv / df_dv
    dtT_df = dtT_dv / df_dv

    Atilde_B = A_time * np.sqrt(np.abs(dtB_df))
    Atilde_T = A_time * np.sqrt(np.abs(dtT_df))

    psiB = 2.0 * np.pi * f * tB - PhiB - np.pi / 4.0
    psiT = 2.0 * np.pi * f * tT - PhiT - np.pi / 4.0

    hBf = Atilde_B * np.exp(1j * psiB)
    hTf = Atilde_T * np.exp(1j * psiT)
    hRf = hBf - hTf

    return {
        "f": f,
        "v": v,
        "tB": tB,
        "tT": tT,
        "PhiB": PhiB,
        "PhiT": PhiT,
        "deltaPhi": PhiB - PhiT,
        "Lambda": Lambda_of_v(v, p),
        "FhatPP": Fhat_PP(v, p.a),
        "hBf": hBf,
        "hTf": hTf,
        "hRf": hRf,
    }


# ============================================================
# inner products and match
# ============================================================
def inner_product(h1f, h2f, f, Sn):
    return 4.0 * np.real(np.trapz(h1f * np.conj(h2f) / Sn, f))


def snr(hf, f, Sn):
    val = inner_product(hf, hf, f, Sn)
    return np.sqrt(max(val, 0.0))


def maximized_match_tc_phi(h1f, h2f, f, Sn, tc_days=0.05, ntc=1201):
    tc_grid = np.linspace(-tc_days * 86400.0, tc_days * 86400.0, ntc)

    n1 = np.sqrt(inner_product(h1f, h1f, f, Sn))
    n2 = np.sqrt(inner_product(h2f, h2f, f, Sn))
    if n1 <= 0 or n2 <= 0:
        return 0.0, 0.0, 0.0

    best_match = -np.inf
    best_tc = 0.0
    best_phase = 0.0

    for tc in tc_grid:
        shifted = h2f * np.exp(1j * 2.0 * np.pi * f * tc)
        z = 4.0 * np.trapz(h1f * np.conj(shifted) / Sn, f)
        m = np.abs(z) / (n1 * n2)
        if m > best_match:
            best_match = m
            best_tc = tc
            best_phase = -np.angle(z)

    return float(best_match), float(best_tc), float(best_phase)


# ============================================================
# calibration helpers
# ============================================================
def end_dephasing_rad(p: Params):
    return float(build_branches(p)["deltaPhi"][-1])


def calibrate_kappa0_for_target_dephasing(p: Params, target=5.0e3, low=1e2, high=1e9, niter=50):
    plo = replace(p, kappa0=low)
    phi_lo = end_dephasing_rad(plo)
    phi_hi = end_dephasing_rad(replace(p, kappa0=high))

    if not (phi_lo <= target <= phi_hi):
        return p

    a, b = low, high
    for _ in range(niter):
        mid = np.sqrt(a * b)
        pmid = replace(p, kappa0=mid)
        phi_mid = end_dephasing_rad(pmid)
        if phi_mid < target:
            a = mid
        else:
            b = mid

    return replace(p, kappa0=np.sqrt(a * b))


def calibrate_distance_for_target_rhoB(p: Params, target_rhoB=5.064):
    data = build_branches(p)
    f = data["f"]
    rho_now = snr(data["hBf"], f, lisa_psd(f))
    if rho_now <= 0:
        return p
    return replace(p, DL_Gpc=p.DL_Gpc * (rho_now / target_rhoB))


def prepare_benchmark():
    p = Params()
    p = calibrate_kappa0_for_target_dephasing(p, target=5.0e3)
    p = calibrate_distance_for_target_rhoB(p, target_rhoB=5.064)
    return p


# ============================================================
# Fisher tools
# ============================================================
def waveform_baseline_only(p: Params):
    data = build_branches(p)
    return data["f"], data["hBf"]


def finite_difference_derivative(p: Params, par_name: str, frac_eps: float):
    if par_name == "lnM":
        p_plus = replace(p, M=p.M * np.exp(frac_eps))
        p_minus = replace(p, M=p.M * np.exp(-frac_eps))
        denom = 2.0 * frac_eps
    elif par_name == "lnmu":
        p_plus = replace(p, mu=p.mu * np.exp(frac_eps))
        p_minus = replace(p, mu=p.mu * np.exp(-frac_eps))
        denom = 2.0 * frac_eps
    elif par_name == "a":
        p_plus = replace(p, a=p.a + frac_eps)
        p_minus = replace(p, a=p.a - frac_eps)
        denom = 2.0 * frac_eps
    else:
        raise ValueError("Unsupported parameter")

    f_plus, h_plus = waveform_baseline_only(p_plus)
    f_minus, h_minus = waveform_baseline_only(p_minus)

    fmin = max(f_plus.min(), f_minus.min())
    fmax = min(f_plus.max(), f_minus.max())
    f = np.linspace(fmin, fmax, min(len(f_plus), len(f_minus)))

    hpr = interp1d(f_plus, h_plus, kind="linear", fill_value=0.0, bounds_error=False)(f)
    hmr = interp1d(f_minus, h_minus, kind="linear", fill_value=0.0, bounds_error=False)(f)
    dh = (hpr - hmr) / denom
    return f, dh


def fisher_bias_analysis(p: Params):
    data = build_branches(p)
    f_full = data["f"]
    hB_full = data["hBf"]
    hT_full = data["hTf"]

    names = ["lnM", "lnmu", "a"]
    eps_map = {"lnM": p.eps_lnM, "lnmu": p.eps_lnmu, "a": p.eps_a}

    derivs = {}
    for nm in names:
        derivs[nm] = finite_difference_derivative(p, nm, eps_map[nm])

    fmin = max([f_full.min()] + [derivs[nm][0].min() for nm in names])
    fmax = min([f_full.max()] + [derivs[nm][0].max() for nm in names])
    f = np.linspace(fmin, fmax, 3000)
    Sn = lisa_psd(f)

    hB = interp1d(f_full, hB_full, kind="linear", fill_value=0.0, bounds_error=False)(f)
    hT = interp1d(f_full, hT_full, kind="linear", fill_value=0.0, bounds_error=False)(f)
    delta_h = hT - hB

    dlist = []
    for nm in names:
        fd, dwdp = derivs[nm]
        d_interp = interp1d(fd, dwdp, kind="linear", fill_value=0.0, bounds_error=False)(f)
        dlist.append(d_interp)

    npar = len(names)
    Gamma = np.zeros((npar, npar))
    bvec = np.zeros(npar)

    for i in range(npar):
        for j in range(npar):
            Gamma[i, j] = inner_product(dlist[i], dlist[j], f, Sn)
        bvec[i] = inner_product(delta_h, dlist[i], f, Sn)

    Cov = inv(Gamma)
    bias = Cov @ bvec
    sigma = np.sqrt(np.diag(Cov))

    return {
        "names": names,
        "f": f,
        "Cov": Cov,
        "Gamma": Gamma,
        "bias": bias,
        "sigma": sigma,
        "bias_over_sigma": bias / sigma,
    }


def draw_cov_ellipse(ax, mean, cov2, nsig=1.0, label=None):
    vals, vecs = eig(cov2)
    order = np.argsort(vals)[::-1]
    vals = np.real(vals[order])
    vecs = np.real(vecs[:, order])

    width = 2.0 * nsig * np.sqrt(max(vals[0], 0.0))
    height = 2.0 * nsig * np.sqrt(max(vals[1], 0.0))
    angle = np.degrees(np.arctan2(vecs[1, 0], vecs[0, 0]))

    ell = plt.matplotlib.patches.Ellipse(
        xy=mean,
        width=width,
        height=height,
        angle=angle,
        fill=False,
        lw=1.8,
        label=label
    )
    ax.add_patch(ell)
```

```python
# ============================================================
# fig:refit_mismatch
# ============================================================
def fig_refit_mismatch(
    p: Params,
    outfile="fig_refit_mismatch.pdf",
    plane="M-a",
    ngrid=41
):
    """
    plane:
        "M-a"  -> mismatch in (M, a)
        "M-mu" -> mismatch in (M, mu)
    """
    data_true = build_branches(p)
    f_true = data_true["f"]
    h_true = data_true["hTf"]

    if plane == "M-a":
        xvals = np.linspace(0.96 * p.M, 1.04 * p.M, ngrid)
        yvals = np.linspace(max(0.5, p.a - 0.08), min(0.999, p.a + 0.08), ngrid)
        xlabel = r"$M\,[10^5 M_\odot]$"
        ylabel = r"$\hat a$"
    elif plane == "M-mu":
        xvals = np.linspace(0.96 * p.M, 1.04 * p.M, ngrid)
        yvals = np.linspace(0.85 * p.mu, 1.15 * p.mu, ngrid)
        xlabel = r"$M\,[10^5 M_\odot]$"
        ylabel = r"$\mu\,[M_\odot]$"
    else:
        raise ValueError("plane must be 'M-a' or 'M-mu'")

    mismatch = np.full((len(yvals), len(xvals)), np.nan)
    best = {"match": -np.inf, "x": None, "y": None, "tc": None, "phi": None}

    for iy, y in enumerate(yvals):
        for ix, x in enumerate(xvals):
            if plane == "M-a":
                pb = replace(p, M=float(x), a=float(y))
            else:
                pb = replace(p, M=float(x), mu=float(y))

            model = build_branches(pb)

            fmin = max(f_true.min(), model["f"].min())
            fmax = min(f_true.max(), model["f"].max())
            if fmax <= fmin:
                continue

            f = np.linspace(fmin, fmax, 2500)
            Sn = lisa_psd(f)

            hT = interp1d(f_true, h_true, kind="linear", fill_value=0.0, bounds_error=False)(f)
            hB = interp1d(model["f"], model["hBf"], kind="linear", fill_value=0.0, bounds_error=False)(f)

            match, tc_best, phi_best = maximized_match_tc_phi(
                hT, hB, f, Sn, tc_days=p.match_tc_days, ntc=p.ntc
            )
            mm = 1.0 - match
            mismatch[iy, ix] = mm

            if match > best["match"]:
                best = {"match": match, "x": x, "y": y, "tc": tc_best, "phi": phi_best}

    fig, ax = plt.subplots(figsize=(6.2, 5.1))

    extent = [
        xvals[0] / 1e5, xvals[-1] / 1e5,
        yvals[0], yvals[-1]
    ]

    im = ax.imshow(
        mismatch,
        origin="lower",
        aspect="auto",
        extent=extent
    )
    cbar = plt.colorbar(im, ax=ax)
    cbar.set_label(r"Mismatch  $1-\mathrm{Match}$")

    if plane == "M-a":
        ax.scatter([p.M / 1e5], [p.a], marker="x", s=70, label="transition benchmark")
    else:
        ax.scatter([p.M / 1e5], [p.mu], marker="x", s=70, label="transition benchmark")

    ax.scatter(
        [best["x"] / 1e5],
        [best["y"]],
        marker="o",
        s=50,
        facecolors="none",
        label="best baseline refit"
    )

    ax.set_xlabel(xlabel)
    ax.set_ylabel(ylabel)
    
    ax.legend(loc="best")

    txt = (
        rf"$\mathcal{{M}}_{{\min}} = {1.0 - best['match']:.3e}$" + "\n" +
        rf"$t_c^* = {best['tc']:.2f}\,\mathrm{{s}}$"
    )
    ax.text(
        0.03, 0.97, txt,
        transform=ax.transAxes,
        va="top",
        ha="left",
        bbox=dict(boxstyle="round", facecolor="white", alpha=0.8)
    )

    fig.tight_layout()
    fig.savefig(outfile, bbox_inches="tight")
    plt.close(fig)

    return best, mismatch
```

```python
# ============================================================
# fig:fisher_bias
# ============================================================
def fig_fisher_bias(
    p: Params,
    outfile="fig_fisher_bias.pdf"
):
    res = fisher_bias_analysis(p)

    names = res["names"]
    bias = res["bias"]
    sigma = res["sigma"]
    bias_over_sigma = np.abs(res["bias_over_sigma"])
    Cov = res["Cov"]

    pretty = [r"$\ln M$", r"$\ln \mu$", r"$\hat a$"]

    fig = plt.figure(figsize=(10.6, 4.2))

    # --------------------------------
    # panel (a): bias significance
    # --------------------------------
    ax1 = fig.add_subplot(1, 3, 1)
    x = np.arange(len(names))
    ax1.bar(x, bias_over_sigma)
    ax1.axhline(1.0, ls="--", lw=1.2)
    ax1.set_xticks(x)
    ax1.set_xticklabels(pretty)
    ax1.set_ylabel(r"$|\Delta \theta_{\rm sys}|/\sigma$")
    
    # --------------------------------
    # panel (b): (ln M, a) ellipse
    # --------------------------------
    ax2 = fig.add_subplot(1, 3, 2)
    idx_M, idx_a = 0, 2
    cov2 = Cov[np.ix_([idx_M, idx_a], [idx_M, idx_a])]
    draw_cov_ellipse(ax2, np.array([0.0, 0.0]), cov2, nsig=1.0, label=r"$1\sigma$")
    draw_cov_ellipse(ax2, np.array([0.0, 0.0]), cov2, nsig=2.0, label=r"$2\sigma$")
    ax2.arrow(
        0.0, 0.0, bias[idx_M], bias[idx_a],
        width=0.0,
        length_includes_head=True
    )
    ax2.set_xlabel(r"$\Delta \ln M$")
    ax2.set_ylabel(r"$\Delta \hat a$")
    
    ax2.legend(loc="best")

    # --------------------------------
    # panel (c): (ln M, ln mu) ellipse
    # --------------------------------
    ax3 = fig.add_subplot(1, 3, 3)
    idx_M, idx_mu = 0, 1
    cov2b = Cov[np.ix_([idx_M, idx_mu], [idx_M, idx_mu])]
    draw_cov_ellipse(ax3, np.array([0.0, 0.0]), cov2b, nsig=1.0, label=r"$1\sigma$")
    draw_cov_ellipse(ax3, np.array([0.0, 0.0]), cov2b, nsig=2.0, label=r"$2\sigma$")
    ax3.arrow(
        0.0, 0.0, bias[idx_M], bias[idx_mu],
        width=0.0,
        length_includes_head=True
    )
    ax3.set_xlabel(r"$\Delta \ln M$")
    ax3.set_ylabel(r"$\Delta \ln \mu$")
    
    ax3.legend(loc="best")

    fig.tight_layout()
    fig.savefig(outfile, bbox_inches="tight")
    plt.close(fig)

    return res
```

```python
# ============================================================
# fig:window_scan
# ============================================================
def fig_window_scan(
    p: Params,
    outfile="fig_window_scan.pdf",
    ncenter=34,
    nwidth=30
):
    centers = np.linspace(0.011, 0.024, ncenter)
    widths = np.linspace(0.003, 0.012, nwidth)

    MM = np.full((len(widths), len(centers)), np.nan)
    RR = np.full((len(widths), len(centers)), np.nan)
    DP = np.full((len(widths), len(centers)), np.nan)

    for iw, w in enumerate(widths):
        for ic, fc in enumerate(centers):
            f_h = fc - 0.5 * w
            f_q = fc + 0.5 * w

            if f_h <= p.f_start or f_q >= p.f_cut or f_q <= f_h:
                continue

            ps = replace(p, f_h=float(f_h), f_q=float(f_q))
            data = build_branches(ps)
            f = data["f"]
            Sn = lisa_psd(f)

            rhoR = snr(data["hRf"], f, Sn)
            match, _, _ = maximized_match_tc_phi(
                data["hTf"], data["hBf"], f, Sn,
                tc_days=p.match_tc_days,
                ntc=max(401, p.ntc // 2)
            )

            MM[iw, ic] = 1.0 - match
            RR[iw, ic] = rhoR
            DP[iw, ic] = data["deltaPhi"][-1]

    fig = plt.figure(figsize=(12.6, 4.0))

    # mismatch
    ax1 = fig.add_subplot(1, 3, 1)
    im1 = ax1.imshow(
        MM,
        origin="lower",
        aspect="auto",
        extent=[centers[0], centers[-1], widths[0], widths[-1]]
    )
    cbar1 = plt.colorbar(im1, ax=ax1)
    cbar1.set_label(r"Mismatch  $\mathcal{M}$")
    ax1.set_xlabel(r"window center $f_c$ [Hz]")
  
    

    # residual SNR
    ax2 = fig.add_subplot(1, 3, 2)
    im2 = ax2.imshow(
        RR,
        origin="lower",
        aspect="auto",
        extent=[centers[0], centers[-1], widths[0], widths[-1]]
    )
    cbar2 = plt.colorbar(im2, ax=ax2)
    cbar2.set_label(r"Residual SNR  $\rho_{\rm R}$")
    ax2.set_xlabel(r"window center $f_c$ [Hz]")


    # phase drift
    ax3 = fig.add_subplot(1, 3, 3)
    im3 = ax3.imshow(
        DP,
        origin="lower",
        aspect="auto",
        extent=[centers[0], centers[-1], widths[0], widths[-1]]
    )
    cbar3 = plt.colorbar(im3, ax=ax3)
    cbar3.set_label(r"$\Delta\Phi_{22}^{\rm SF}$ [rad]")
    ax3.set_xlabel(r"window center $f_c$ [Hz]")
 

    fig.tight_layout()
    fig.savefig(outfile, bbox_inches="tight")
    plt.close(fig)

    return {"centers": centers, "widths": widths, "mismatch": MM, "rhoR": RR, "dphi": DP}
```
